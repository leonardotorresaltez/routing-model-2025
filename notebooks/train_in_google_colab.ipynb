Step 1 - configure branch to run GITHUB-BRANCH

```python
GITHUB_BRANCH = "main"  # <--- change it : "develop", "feature-x", etc.
REPO_URL = "https://github.com/leonardotorresaltez/routing-model-2025"
REPO_NAME = "routing-model-2025"
```

Step 2 - clone repo

```python
import os
import subprocess

# 1. Clonar o actualizar el repositorio según la rama
if not os.path.exists(f'/content/{REPO_NAME}'):
    print(f"Clonando rama '{GITHUB_BRANCH}'...")
    !git clone -b {GITHUB_BRANCH} {REPO_URL}
else:
    print("El repositorio ya existe. Actualizando rama...")
    %cd /content/{REPO_NAME}
    !git checkout {GITHUB_BRANCH}
    !git pull origin {GITHUB_BRANCH}

%cd /content/{REPO_NAME}
```

Step 3 - install poetry and dependencies

```python
# 2. Instalación condicional de Poetry
try:
    import poetry
    print("Poetry ya está instalado.")
except ImportError:
    if not os.path.exists('/root/.local/bin/poetry'):
        print("Instalando Poetry por primera vez...")
        !curl -sSL https://install.python-poetry.org | python3 -
    else:
        print("Poetry ya existe en el sistema.")

# 3. Asegurar PATH y Configuración
os.environ["PATH"] += ":/root/.local/bin"
!poetry config virtualenvs.create false

# 4. Instalar dependencias (Poetry saltará lo que ya esté listo)
print("Sincronizando dependencias...")
!poetry install --no-interaction --no-ansi

print("\n✅ Entorno listo para entrenar.")
```
